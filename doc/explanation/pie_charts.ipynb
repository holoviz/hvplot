# When to Use Pie Charts

Pie charts have a bad reputation among data visualization practitioners, and it is partly deserved: people generally judge angles and areas less precisely than lengths, so a pie is rarely the best chart to compare values. Yet pies are also one of the most familiar charts, and for the right job they work well. This guide gives practical advice on when to use a [pie](../ref/api/manual/hvplot.hvPlot.pie.ipynb) or [donut](../ref/api/manual/hvplot.hvPlot.donut.ipynb) chart, and when a bar chart or another chart is a better choice.

## What pie charts are good at

A pie chart answers one question well: *how big is each part of the whole?* It is a good choice when:

- the parts add up to a meaningful whole, like a budget, the respondents of a survey or the individuals of a population;
- there are only a few categories, roughly 2 to 5;
- the main message is a single part-to-whole statement, such as "almost half" or "about a fifth";
- the audience reads pies easily, which is often the case for a general audience.

The pie below makes its point at a glance: Adelie penguins make up almost half of the penguins of the dataset, and Chinstrap penguins about a fifth.

```python
import hvplot.pandas  # noqa

df = hvplot.sampledata.penguins('pandas')
counts = df['species'].value_counts()

counts.hvplot.pie(
    show_labels='{species}\n{fraction:.0%}',
    legend=False,
    title='Penguins by species',
)
```

## Where pie charts struggle

Pie charts are a weaker choice for:

- **Comparing similar values.** Slices of 20% and 23% look about the same, and ranking them by eye is hard. Bars sharing a common baseline make small differences visible.
- **Many categories.** Beyond five or so categories, the slices get thin, the labels collide and the colors become hard to tell apart and to match with the legend.
- **Comparing several pies.** Comparing two pies side by side means matching colors and judging angles across charts, which makes differences between groups hard to see.
- **Values that don't add up to a meaningful whole.** Averages, rates, or the answers to a survey question accepting several choices don't make a whole, and a pie suggests they do.
- **Negative values.** A slice can't be negative, hvPlot raises an error if it finds negative values.
- **Change over time.** A series of pies is a poor way to show a trend, a line or bar chart shows it directly.

The example below shows the sales of five regions with close values. In the pie, it is hard to tell which region sells the most, while the bar chart makes it obvious.

```python
import pandas as pd
import hvplot.pandas  # noqa

df = pd.DataFrame({
    'Region': ['North', 'East', 'South', 'West', 'Central'],
    'Sales': [20, 23, 17, 21, 19],
})

pie = df.hvplot.pie(x='Region', y='Sales', title='Pie', frame_width=250)
bar = df.hvplot.bar(x='Region', y='Sales', title='Bar', frame_width=250)
pie + bar
```

## Tips for better pie charts

When a pie is the right choice, these tips help make it easy to read:

- **Aggregate first.** hvPlot draws one slice per row and doesn't sum the rows sharing the same category. Aggregate the data first, for example with `value_counts()` or `groupby(...).sum()`.
- **Group small slices into "Other".** A few large slices and an "Other" slice are easier to read than many thin slices.
- **Order the slices.** Sorting the data, largest first, makes the slices easier to compare.
- **Label with percentages.** `show_labels='{fraction:.0%}'` draws the share of each slice, and a template like `'{Expense}\n{fraction:.0%}'` adds the category, which lets you hide the legend with `legend=False`.
- **Limit the number of colors.** Fewer slices means fewer colors to tell apart. Pass a dictionary to `cmap` to keep the same color for a category across charts.
- **Use the donut's hole.** The hole of a [donut](../ref/api/manual/hvplot.hvPlot.donut.ipynb) chart keeps the focus on the ring, and leaves room to display a total, set with `.opts(center_label=...)`. Its size is set with `inner_radius`.
- **Avoid 3D and exploded slices.** They distort the angles and areas the reader relies on, without adding information. hvPlot doesn't draw them.

The example below applies most of these tips to a list of monthly expenses: the amounts are summed per expense, sorted, the three smallest expenses are grouped into "Other", the slices are labeled with their share, and the total is displayed in the hole.

```python
import pandas as pd
import hvplot.pandas  # noqa

transactions = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Food', 'Leisure', 'Transport', 'Food', 'Phone', 'Gifts'],
    'Amount': [1250, 120, 95, 250, 150, 140, 40, 30],
})

totals = transactions.groupby('Expense')['Amount'].sum().sort_values(ascending=False)
totals = pd.concat([totals.head(3), pd.Series({'Other': totals.iloc[3:].sum()})])
totals = totals.rename_axis('Expense').reset_index(name='Amount')

totals.hvplot.donut(
    x='Expense',
    y='Amount',
    show_labels='{Expense}\n{fraction:.0%}',
    legend=False,
    title='Monthly expenses',
).opts(center_label='Total\n{total:,.0f}')
```

## Alternatives

When a pie isn't the right fit, these charts often are:

- **Bar charts**, with {meth}`bar <hvplot.hvPlot.bar>` or {meth}`barh <hvplot.hvPlot.barh>`, to compare values precisely or to show many categories. Horizontal bars leave more room for long category names.
- **Stacked bar charts**, with `stacked=True`, to show how a whole is split across several groups. Each bar is a whole, and the bars share a common axis, which makes them easier to compare than several pies.
- **Line charts**, with {meth}`line <hvplot.hvPlot.line>`, to show change over time.

The stacked bar chart below compares how a monthly budget is split between expenses in two years, a job that would otherwise need two pies.

```python
import pandas as pd
import hvplot.pandas  # noqa

df = pd.DataFrame({
    'Year': ['2023'] * 4 + ['2024'] * 4,
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure'] * 2,
    'Amount': [1100, 400, 150, 300, 1200, 450, 180, 250],
})

df.hvplot.barh(x='Year', y='Amount', by='Expense', stacked=True, legend='right')
```

## Summary

Use a pie or donut chart to show how a meaningful whole splits into a few parts, when one part-to-whole message matters most. Prefer a bar chart to compare values, especially close ones, or when there are many categories, and a stacked bar chart to compare proportions across groups.
