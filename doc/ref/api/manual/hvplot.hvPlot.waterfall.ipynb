## Examples

### Basic waterfall chart

A waterfall chart shows how a starting value is changed by a series of increases and decreases. Each bar starts at the running total of the previous bars and a final bar shows the total.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Step': ['Revenue', 'Cost of sales', 'Marketing', 'Salaries', 'Other income', 'Taxes'],
    'Change': [1000, -350, -120, -280, 90, -85],
})

df.hvplot.waterfall(x='Step', y='Change', title='Profit and loss', width=600)
````

### Colors and total

The bars are colored by the direction of their change with `positive_color` and `negative_color`, while `start_color` and `total_color` set the color of the first and last bars. The total bar can be renamed with `total_label` or removed with `show_total=False`.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Step': ['Revenue', 'Cost of sales', 'Marketing', 'Salaries', 'Other income', 'Taxes'],
    'Change': [1000, -350, -120, -280, 90, -85],
})

df.hvplot.waterfall(
    x='Step', y='Change', width=600, total_label='Net profit',
    positive_color='#2a9d8f', negative_color='#e76f51',
    start_color='#264653', total_color='#e9c46a',
)
````

### Horizontal waterfall chart

Set `invert=True` to draw the bars horizontally, here without the lines connecting them.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Step': ['Revenue', 'Cost of sales', 'Marketing', 'Salaries', 'Other income', 'Taxes'],
    'Change': [1000, -350, -120, -280, 90, -85],
})

df.hvplot.waterfall(x='Step', y='Change', invert=True, show_connectors=False, height=350)
````

### Matplotlib backend

````python
import hvplot.pandas  # noqa
import pandas as pd
hvplot.extension('matplotlib')

df = pd.DataFrame({
    'Step': ['Revenue', 'Cost of sales', 'Marketing', 'Salaries', 'Other income', 'Taxes'],
    'Change': [1000, -350, -120, -280, 90, -85],
})

df.hvplot.waterfall(x='Step', y='Change', rot=30)
````
