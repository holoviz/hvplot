## Examples

A pie chart shows how a whole is split into parts. To draw a ring with a hole in the center, see {meth}`hvplot.hvPlot.donut`.

### Basic pie chart

The categories of the wedges are drawn from `x` and their sizes from `y`.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.pie(x='Expense', y='Amount')
````

### Pie chart from a Series

As in Pandas, when `x` is not set the index is used for the categories, which makes it easy to plot a Series.

````python
import hvplot.pandas  # noqa
import pandas as pd

budget = pd.Series(
    [1200, 450, 180, 250, 400],
    index=pd.Index(['Rent', 'Food', 'Transport', 'Leisure', 'Savings'], name='Expense'),
    name='Amount',
)

budget.hvplot.pie()
````

### Counting categories

Each row is drawn as a wedge, values of rows sharing the same category are not summed. To show how often each category appears in a dataset, count the rows first, for example with `value_counts`.

````python
import hvplot.pandas  # noqa

df = hvplot.sampledata.penguins('pandas')

df['species'].value_counts().hvplot.pie()
````

### Wedge labels

Set `show_labels=True` to draw the category of each wedge on the wedge itself. `show_labels` also accepts a template string formatted with the field names and the `fraction` of the total, here used to display percentages.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.pie(x='Expense', y='Amount', show_labels='{fraction:.0%}')
````

### Wedge colors

The wedges are colored by category with `cmap`, which accepts a colormap name, a list of colors or a dictionary mapping each category to a color.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.pie(
    x='Expense',
    y='Amount',
    cmap={
        'Rent': 'firebrick',
        'Food': 'orange',
        'Transport': 'gold',
        'Leisure': 'teal',
        'Savings': 'seagreen',
    },
)
````

### One pie per group

Grouping the data with `by` lays out one pie per group.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Year': [2023] * 4 + [2024] * 4,
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure'] * 2,
    'Amount': [1100, 400, 150, 300, 1200, 450, 180, 250],
})

df.hvplot.pie(x='Expense', y='Amount', by='Year', frame_width=250)
````

### Matplotlib backend

Switch to Matplotlib with `hvplot.extension('matplotlib')`.

````python
import hvplot.pandas  # noqa
import pandas as pd
hvplot.extension('matplotlib')

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.pie(x='Expense', y='Amount', show_labels='{fraction:.0%}')
````
