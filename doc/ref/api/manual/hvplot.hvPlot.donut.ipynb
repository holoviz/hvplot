## Examples

A donut chart is a {meth}`pie <hvplot.hvPlot.pie>` chart with a hole in its center. It accepts the same options as `pie`, the examples below focus on the ones specific to donuts.

### Basic donut chart

The categories of the wedges are drawn from `x` (or the index if not set) and their sizes from `y`.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.donut(x='Expense', y='Amount')
````

### Hole size

`inner_radius` sets the size of the hole, as a fraction of the outer radius between 0 and 1. The default is 0.5.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

plot_opts = dict(x='Expense', y='Amount', frame_width=200, legend=False)
(
    df.hvplot.donut(inner_radius=0.3, title='inner_radius=0.3', **plot_opts)
    + df.hvplot.donut(inner_radius=0.6, title='inner_radius=0.6', **plot_opts)
    + df.hvplot.donut(inner_radius=0.9, title='inner_radius=0.9', **plot_opts)
)
````

### Wedge labels

As with `pie`, `show_labels` draws a label on each wedge, either the category (`True`) or a template string formatted with the field names and the `fraction` of the total.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.donut(x='Expense', y='Amount', show_labels='{fraction:.0%}')
````

### Center label

`center_label` displays a text in the hole: `'total'` displays the sum of the values, and any other string is a template formatted with `total` and the `y` field name, both the sum of the values.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.donut(x='Expense', y='Amount', center_label='Total\n{Amount:,.0f}')
````

### One donut per group

Grouping the data with `by` lays out one donut per group, which is a compact way to compare proportions across groups.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Year': [2023] * 4 + [2024] * 4,
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure'] * 2,
    'Amount': [1100, 400, 150, 300, 1200, 450, 180, 250],
})

df.hvplot.donut(x='Expense', y='Amount', by='Year', frame_width=250)
````

### Matplotlib backend

Switch to Matplotlib with `hvplot.extension('matplotlib')`.

````python
import hvplot.pandas  # noqa
import pandas as pd
hvplot.extension('matplotlib')

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.donut(x='Expense', y='Amount', show_labels='{fraction:.0%}')
````
