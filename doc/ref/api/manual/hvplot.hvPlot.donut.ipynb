## Examples

### Basic donut chart

A donut chart shows how a whole is split into parts. The categories are drawn from `x` and the sizes of the wedges from `y`.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.donut(x='Expense', y='Amount', title='Monthly budget')
````

### Pie chart

Setting `inner_radius` to `0` draws a pie chart. Each wedge can be labelled with `show_labels`, which also accepts a template formatted with the field names and the `fraction` of the total.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.donut(
    x='Expense', y='Amount', inner_radius=0, show_labels='{fraction:.0%}', legend='right'
)
````

### Center label and colors

`center_label` adds a text in the hole of the ring, `'total'` displaying the sum of the values. The wedges are colored by category with `cmap`, which can be a colormap name, a list of colors or a dictionary mapping each category to a color.

````python
import hvplot.pandas  # noqa
import pandas as pd

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.donut(
    x='Expense', y='Amount', inner_radius=0.6, legend=False,
    center_label='Total: {total:,.0f}', cmap='Set2', width=350,
)
````

### One donut per group

As overlaid donuts would hide each other, grouping the data with `by`, or selecting several `y` columns, lays out one donut per group. Each donut maps its own categories to colors, so a dictionary `cmap` keeps the colors consistent across the layout.

````python
import hvplot.pandas  # noqa

df = hvplot.sampledata.penguins('pandas')
counts = df.groupby(['island', 'species'], observed=True).size().reset_index(name='count')

counts.hvplot.donut(
    x='species', y='count', by='island', width=250,
    cmap={'Adelie': '#1f77b4', 'Chinstrap': '#ff7f0e', 'Gentoo': '#2ca02c'},
)
````

### Matplotlib backend

````python
import hvplot.pandas  # noqa
import pandas as pd
hvplot.extension('matplotlib')

df = pd.DataFrame({
    'Expense': ['Rent', 'Food', 'Transport', 'Leisure', 'Savings'],
    'Amount': [1200, 450, 180, 250, 400],
})

df.hvplot.donut(x='Expense', y='Amount', inner_radius=0.5, show_labels=True, legend=False)
````
