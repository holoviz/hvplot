# Pie Chart with a Groupby Widget

A pie chart of the number of penguins of each species, with a widget to select the island. Setting `groupby` creates one pie per island and displays only the selected one. The static rendering below shows the first island. A `cmap` dictionary keeps the color of each species the same across islands.

```python
import hvplot.pandas  # noqa

df = hvplot.sampledata.penguins('pandas')
counts = df.groupby(['island', 'species']).size().reset_index(name='count')
colors = {'Adelie': '#1f77b4', 'Chinstrap': '#2ca02c', 'Gentoo': '#ff7f0e'}

counts.hvplot.pie(
    x='species', y='count', groupby='island', cmap=colors,
    title='Pie Chart with a Groupby Widget (Bokeh)',
)
```

```python
import hvplot.pandas  # noqa
hvplot.extension('matplotlib')

df = hvplot.sampledata.penguins('pandas')
counts = df.groupby(['island', 'species']).size().reset_index(name='count')
colors = {'Adelie': '#1f77b4', 'Chinstrap': '#2ca02c', 'Gentoo': '#ff7f0e'}

counts.hvplot.pie(
    x='species', y='count', groupby='island', cmap=colors,
    title='Pie Chart with a Groupby Widget (Matplotlib)',
)
```
