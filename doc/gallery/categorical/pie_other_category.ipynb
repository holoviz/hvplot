# Pie Chart with an Other Category

A pie chart of the countries of the 2024 earthquakes of the `earthquakes` dataset. The countries counted with `value_counts()` beyond the three most frequent are grouped into an `Other` wedge, which keeps the pie readable.

```python
import pandas as pd
import hvplot.pandas  # noqa

df = hvplot.sampledata.earthquakes('pandas')
countries = df['place'].str.split(', ').str[-1]
counts = countries.value_counts()
counts = pd.concat([counts.head(3), pd.Series({'Other': counts.iloc[3:].sum()})])
counts = counts.rename_axis('country').reset_index(name='count')

counts.hvplot.pie(
    x='country', y='count', show_labels='{fraction:.0%}',
    title='Pie Chart with an Other Category (Bokeh)',
)
```

```python
import pandas as pd
import hvplot.pandas  # noqa
hvplot.extension('matplotlib')

df = hvplot.sampledata.earthquakes('pandas')
countries = df['place'].str.split(', ').str[-1]
counts = countries.value_counts()
counts = pd.concat([counts.head(3), pd.Series({'Other': counts.iloc[3:].sum()})])
counts = counts.rename_axis('country').reset_index(name='count')

counts.hvplot.pie(
    x='country', y='count', show_labels='{fraction:.0%}',
    title='Pie Chart with an Other Category (Matplotlib)',
)
```
