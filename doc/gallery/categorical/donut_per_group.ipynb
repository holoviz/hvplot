# Donut Chart per Group

One donut chart per year showing how a monthly budget is split between expenses. Grouping with `by` lays out one chart per group.

```python
import pandas as pd
import hvplot.pandas  # noqa

df = pd.DataFrame({
    'year': [2023] * 4 + [2024] * 4,
    'expense': ['Rent', 'Food', 'Transport', 'Leisure'] * 2,
    'amount': [1100, 400, 150, 300, 1200, 450, 180, 250],
})

df.hvplot.donut(x='expense', y='amount', by='year', frame_width=250)
```

```python
import pandas as pd
import hvplot.pandas  # noqa
hvplot.extension('matplotlib')

df = pd.DataFrame({
    'year': [2023] * 4 + [2024] * 4,
    'expense': ['Rent', 'Food', 'Transport', 'Leisure'] * 2,
    'amount': [1100, 400, 150, 300, 1200, 450, 180, 250],
})

df.hvplot.donut(x='expense', y='amount', by='year', frame_width=250)
```
