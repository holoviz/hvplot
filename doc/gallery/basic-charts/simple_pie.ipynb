# Simple Pie Chart

A basic pie chart showing the share of each penguin species in the `penguins` dataset. As with pandas, plotting a Series uses its index for the wedge labels and its values for the wedge sizes.

```python
import hvplot.pandas  # noqa

df = hvplot.sampledata.penguins('pandas')
counts = df['species'].value_counts()

counts.hvplot.pie(title='Simple Pie Chart (Bokeh)')
```

```python
import hvplot.pandas  # noqa
hvplot.extension('matplotlib')

df = hvplot.sampledata.penguins('pandas')
counts = df['species'].value_counts()

counts.hvplot.pie(title='Simple Pie Chart (Matplotlib)')
```
