# Simple Donut Chart

A donut chart, a pie chart with a hole in its center, showing the share of each penguin species in the `penguins` dataset. This example uses a DataFrame, with `x` setting the column of the wedge labels and `y` the column of the wedge sizes.

```python
import hvplot.pandas  # noqa

df = hvplot.sampledata.penguins('pandas')
counts = df.groupby('species').size().reset_index(name='count')

counts.hvplot.donut(x='species', y='count', title='Simple Donut Chart (Bokeh)')
```

```python
import hvplot.pandas  # noqa
hvplot.extension('matplotlib')

df = hvplot.sampledata.penguins('pandas')
counts = df.groupby('species').size().reset_index(name='count')

counts.hvplot.donut(x='species', y='count', title='Simple Donut Chart (Matplotlib)')
```
