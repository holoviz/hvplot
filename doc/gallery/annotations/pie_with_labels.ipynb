# Pie Chart with Percentage Labels

A pie chart of the approximate world electricity generation mix in 2023, with each wedge labeled with its source and its share of the total. The labels replace the legend, which is hidden.

```python
import pandas as pd
import hvplot.pandas  # noqa

df = pd.DataFrame({
    'source': ['Coal', 'Gas', 'Hydro', 'Other'],
    'share': [35, 23, 14, 28],
})

df.hvplot.pie(
    x='source',
    y='share',
    show_labels='{source}\n{fraction:.0%}',
    legend=False,
    title='Pie Chart with Percentage Labels (Bokeh)',
)
```

```python
import pandas as pd
import hvplot.pandas  # noqa
hvplot.extension('matplotlib')

df = pd.DataFrame({
    'source': ['Coal', 'Gas', 'Hydro', 'Other'],
    'share': [35, 23, 14, 28],
})

df.hvplot.pie(
    x='source',
    y='share',
    show_labels='{source}\n{fraction:.0%}',
    legend=False,
    title='Pie Chart with Percentage Labels (Matplotlib)',
)
```
